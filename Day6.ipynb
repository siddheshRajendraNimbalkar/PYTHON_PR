```python
# Write a Python program to check if the given number is a Disarium Number

num = str(input("Enter a number: "))

sum = 0

powerNum = 1
for i in num:
    sum += pow(int(i),powerNum)
    powerNum += 1
    
int(num) == sum
```

```python
# Write a Python program to print all disarium numbers between 1 to 100.

for i in range(1,101):
    num = str(i)
    sum = 0
    powerNum = 1
    for j in num:
        sum += pow(int(j),powerNum)
        powerNum += 1
        
    if int(num) == sum :
        print(f"num:{num}")
```

```python
# Write a Python program to check if the given number is Happy Number.

def ReturnPower(num):
    sum = 0
    powerNum = 2
    for i in num:
        sum += pow(int(i),powerNum)
    return int(sum)

def happyNum(num):
    
    seen = list()
    while num != 1:
        num = ReturnPower(str(num))
        if num in seen:
            return False
        seen.append(num)
        if num == 1:
            return True
    return False

happyNum(23)
```

```python
# Write a Python program to print all happy numbers between 1 and 100.

def AllHappyNum():
    for i in range(1,101):
        if happyNum(i):
            print(f"{i}")
AllHappyNum()
```

```python
# Write a Python program to determine whether the given number is a Harshad
# Number.

def hashNum(num):
    sum = 0
    for i in num:
        sum += int(i)

    if int(num)%sum == 0:
        return True
    else:
        return False

num = str(input("Enter a number: "))
hashNum(num)
```

```python
for i in range(1,101):
    num = str(i)
    if hashNum(num):
        print(f"{num}")
```
