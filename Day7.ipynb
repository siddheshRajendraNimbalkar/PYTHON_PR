```python
# Write a Python program to Multiply all numbers in the list.
def ListMult(listOfNum):
    result = 1
    for i in listOfNum:
        result *= i
    return result

ListMult([1,2,3,4,5])
```

```python
# Write a Python program to find smallest number in a list.

def SmallList(lisOfNum):
    return min(lisOfNum)

print(SmallList([1,2,3,4,5]))
print("Without using min function")

def SmallListDif(ListOfNum):
    smallNum = ListOfNum[0]
    for i in range(1,len(ListOfNum)):
        if smallNum > ListOfNum[i]:
            smallNum = ListOfNum[i]
    return smallNum

SmallListDif([2,3,4,5,1])
```

```python
# Write a Python program to find second largest number in a list.
num = [30, 10, 45, 5, 20,50]

def LargestNum(ListOfNum):
    add = []
    count = 0
    for i in range(0,len(ListOfNum)):
        if i == 0:
            add.append(ListOfNum[i])
        elif add[count] < ListOfNum[i]:
            add.append(ListOfNum[i])
            count += 1
        
    return add[-2]
LargestNum(num)
```

```python
# Write a Python program to find N largest elements from a list.

def SortNum(ListOfNum,n):
    sortNum = sorted(ListOfNum)
    NumList = list(dict.fromkeys(sortNum))
    return NumList[n-1:n]

SortNum([30, 10, 45, 5, 20,50],2)          
```
